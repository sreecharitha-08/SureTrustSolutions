```python
import pandas as pd

# Define the sales data
city_sales = {'City': ['New York', 'New York', 'Paris', 'Paris', 'London', 'London', 'New York', 'Paris'], 
               'Product': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'A'], 
               'Sales': [100, 200, 150, 250, 130, 220, 170, 180]}
```

```python

df_sales=pd.DataFrame(city_sales)
city_sales=df_sales.groupby('City')['Sales'].agg(['sum','mean'])
print(city_sales)
```
