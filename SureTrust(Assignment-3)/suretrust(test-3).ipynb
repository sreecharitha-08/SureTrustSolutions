Q1. Write a Python program that takes a sentence as input and returns a dictionary where the
keys are vowels and the values are the number of times each vowel appears in the sentence.
Ignore case.
Example Input: "Education is important"
Expected Output: {'a': 2, 'e': 1, 'i': 2, 'o': 1, 'u': 1}
give simple code for this in python give an optimized one

```python
def count_vowels(sentence):
    sentence = sentence.lower()
    vowels = 'aeiou'
    vowel_count = {v: sentence.count(v) for v in vowels if v in sentence}
    # Ensure the order is a, e, i, o, u
    ordered_vowel_count = {v: vowel_count[v] for v in vowels if v in vowel_count}
    print(ordered_vowel_count)

# Example usage
input_sentence = "Education is important"
count_vowels(input_sentence)
```

Q2. Write a function that checks if a given number is a palindrome. Then write another function
that prints all palindrome numbers between two given numbers (inclusive).
Example Input: start = 100, end = 150
Expected Output: 101 111 121 131 141
give simple code for this in python an optimized one code should give expected output format only

```python
def is_palindrome(number):
    return str(number) == str(number)[::-1]
def print_palindromes(start, end):
    for num in range(start, end + 1):
        if is_palindrome(num):
            print(num, end=' ')
start = 100
end = 150
print_palindromes(start, end)
```

Q3. Create a class Employee with attributes emp_id, name, and salary. Write methods to:
• Display employee details
• Write the details to a file named employee_data.txt in CSV format
• Read from the file and display the content
give simple code with requirements for this code

```python
import csv
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def display(self):
        print(f"ID: {self.emp_id}, Name: {self.name}, Salary: {self.salary}")

    def write_to_file(self, filename='employee_data.txt'):
        with open(filename, 'a', newline='') as file:
            writer = csv.writer(file)
            writer.writerow([self.emp_id, self.name, self.salary])

    @staticmethod
    def read_from_file(filename='employee_data.txt'):
        with open(filename, 'r') as file:
            reader = csv.reader(file)
            for row in reader:
                print(f"ID: {row[0]}, Name: {row[1]}, Salary: {row[2]}")

# Example usage
emp1 = Employee(101, "Alice", 50000)
emp1.display()
emp1.write_to_file()

emp2 = Employee(102, "Bob", 60000)
emp2.write_to_file()

print("\nReading from file:")
Employee.read_from_file()
```

Q4. Create a class LogAnalyzer that reads a file called log.txt, counts the number of lines that contain
the word "error" (case-insensitive), and prints the count. Include proper function(s) and file
handling. Assume log.txt contains multiple lines.
Log.txt:
INFO: System initialized successfully
ERROR: Failed to connect to the database
WARNING: Deprecated function used
info: User logged in
error: Timeout while connecting to server
DEBUG: Memory usage is within limits
ERROR: Null pointer exception occurred
Warning: Battery level low
fatal: System crash detected
Error: File not found
give code for this in python in simple way and an optimized one

```python
import os

class LogAnalyzer:
    def __init__(self, filename='log.txt'):
        self.filename = filename

    def count_errors(self):
        # Debug: Print current working directory
        print("Current working directory:", os.getcwd())
        
        if not os.path.exists(self.filename):  # Check if file exists in current directory
            print(f"File not found: {self.filename}")
            return
        
        count = 0
        with open(self.filename, 'r') as file:
            for line in file:
                if 'error' in line.lower():
                    count += 1
        print(f"Number of lines with 'error': {count}")


# Example usage with relative path (log.txt in the same directory)
analyzer = LogAnalyzer('log.txt')  # Just the file name without path
analyzer.count_errors()
```
