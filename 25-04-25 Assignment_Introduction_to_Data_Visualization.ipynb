# <b><u>Problem 2 </u></b>

### Read face cream and facewash product sales data and show it using the bar chart. The bar chart should display the number of units sold per month for each product. Add a separate bar for each product in the same chart.

### Expected output:


```python
df.groupby(['month_number'])[['facewash','facecream']].mean()


```

```python
bar_width = 0.4
months = df['month_number']
facecream_sales = df['facecream']
facewash_sales = df['facewash']
r1 = months - bar_width/2
r2 = months + bar_width/2
plt.bar(r1, facecream_sales, width=bar_width, label='Face Cream', color='orange')
plt.bar(r2, facewash_sales, width=bar_width, label='Face Wash', color='blue')
plt.xlabel('Month Number')
plt.ylabel('Units Sold')
plt.title('Face Cream and Face Wash Product Sales Month Wise')
plt.legend()
plt.grid(True, axis='y')
plt.show()
```
